````python
# TODO: Step 5 - build a new prompt using an earlier variable
#   (fact, question, answer, and/or the evaluation)
# TODO: call chat(messages, deployment="gpt-5.4-ptu") -- do NOT use the default deployment here
# TODO: extract the result, and print it clearly labeled



from dotenv import load_dotenv
import os
import sys

from openai import AzureOpenAI

load_dotenv(override=True)

api_key = os.getenv("AZURE_APIM_OPENAI_SUBSCRIPTION_KEY")
api_version = os.getenv("AZURE_APIM_OPENAI_API_VERSION")
endpoint = os.getenv("AZURE_APIM_OPENAI_ENDPOINT")
deployment = os.getenv("AZURE_APIM_OPENAI_DEPLOYMENT")

if not all ([api_key, api_version, deployment]):
    sys.exit(
        "Missing Azure APIM settings. Set Subsription, Version, Deployment, and Endpoint in your .env file!"
    )
    

print(f"Azure APIM key exists and begins: {api_key[:6]}")
print(f"Deployment: {deployment}")


openai = AzureOpenAI(
    api_key=api_key,
    api_version=api_version,
    azure_endpoint=endpoint,
)


# chat helper:
def chat(messages, max_completion_tokens=1000):
    return openai.chat.completions.create(
        model=deployment,
        messages=messages,
        max_completion_tokens=max_completion_tokens,
    )

# Fun Fact creation
messages = [{"role": "user", "content": "Tell me a random, short, fun fact about Outer Space"}]
response = chat(messages)
fact = response.choices[0].message.content
print(response.choices[0].message.content)



# BEGIN STEP 5 <<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<

# Evaluation of the Fun Fact and a new Fact that connects to original concept
deployment="gpt-5.4-ptu"
messages = [{"role": "user", "content": response}]

# New Chat helper for gpt-5.4-ptu
def chat(messages, deployment, max_completion_tokens=1000):
    return openai.chat.completions.create(
        model=deployment,
        messages=messages,
        max_completion_tokens=max_completion_tokens,
    )

messages = f"""
Here is a fun fact:
{fact}

Please Create a new fun fact that ties into the original idea of the first. Please keep it fewer than 75 words and interesting. Respond with only your fact.
"""

newMessage = [{"role": "user", "content": messages}]
newFact = chat(newMessage, deployment="gpt-5.4-ptu")
answer = newFact.choices[0].message.content
print("GPT-5.4-PTU Fact:")
print(answer)

from IPython.display import Markdown, display

display(Markdown(answer))
````
